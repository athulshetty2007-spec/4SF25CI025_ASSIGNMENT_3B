```python



df = pd.DataFrame(np.random.randint(1, 100, size=(4, 3)),
                  columns=['A', 'B', 'C'])
print("Row-wise Sum:")
print(df.sum(axis=1))
print("\nColumn-wise Sum:")
print(df.sum(axis=0))
```

```python


# Create DataFrame with 3 columns and 5 rows
df = pd.DataFrame(np.random.randint(1, 100, size=(5, 3)),
                  columns=['A', 'B', 'C'])

df.iloc[1, 0] = np.nan
df.iloc[2, 1] = np.nan
df.iloc[3, 2] = np.nan

print("Original DataFrame:")
print(df)
df = df.fillna(df.mean())

print("\nDataFrame after filling NaN values:")
print(df)
```

```python



df = pd.DataFrame(np.random.randint(1, 100, size=(6, 4)),
                  columns=['A', 'B', 'C', 'D'])
df.iloc[1, 0] = np.nan
df.iloc[3, 2] = np.nan
df.iloc[4, 1] = np.nan

print("Original DataFrame:")
print(df)
df = df.dropna()

print("\nDataFrame after dropping NaN rows:")
print(df)
```

```python



df = pd.DataFrame({
    'Category': np.random.choice(['A', 'B', 'C'], size=10),
    'Value': np.random.randint(1, 100, size=10)
})
print("Original DataFrame:")
print(df)
result = df.groupby('Category')['Value'].agg(['sum', 'mean'])
print("\nSum and Mean of Each Category:")
print(result)
```

```python



df = pd.DataFrame({
    'Product': np.random.choice(['Laptop', 'Mobile', 'Tablet', 'TV'], size=10),
    'Category': np.random.choice(['Electronics', 'Furniture', 'Clothing'], size=10),
    'Sales': np.random.randint(100, 1000, size=10)
})

print("Original DataFrame:")
print(df)
result = df.groupby('Category')['Sales'].sum()
print("\nTotal Sales by Category:")
print(result)
```

```python

df1 = pd.DataFrame({
    'ID': [1, 2, 3, 4],
    'Name': ['Athul', 'Rahul', 'Arun', 'Kiran']
})

df2 = pd.DataFrame({
    'ID': [1, 2, 3, 4],
    'Marks': [85, 90, 75, 88]
})

print("First DataFrame:")
print(df1)

print("\nSecond DataFrame:")
print(df2)


merged_df = pd.merge(df1, df2, on='ID')

print("\nMerged DataFrame:")
print(merged_df)
```

```python



index = pd.MultiIndex.from_tuples(
    [('A', 1), ('A', 2), ('B', 1), ('B', 2)],
    names=['Category', 'Number']
)


df = pd.DataFrame({
    'Value': [10, 20, 30, 40],
    'Marks': [85, 90, 75, 95]
}, index=index)

print("Original MultiIndex DataFrame:")
print(df)

print("\nAccess Category A:")
print(df.loc['A'])
print("\nAccess A, 1:")
print(df.loc[('A', 1)])
print("\nSlicing Category A to B:")
print(df.loc['A':'B'])
```

```python


# Create MultiIndex
index = pd.MultiIndex.from_tuples(
    [('A', 'X'), ('A', 'Y'), ('B', 'X'), ('B', 'Y'), ('C', 'X'), ('C', 'Y')],
    names=['Category', 'SubCategory']
)

# Create DataFrame with random values
df = pd.DataFrame({
    'Value': np.random.randint(1, 100, size=6)
}, index=index)

print("Original DataFrame:")
print(df)

# Compute sum for each Category and SubCategory
result = df.groupby(level=['Category', 'SubCategory'])['Value'].sum()

print("\nSum of Values:")
print(result)
```

```python


# Create DataFrame with 3 columns and 5 rows
df = pd.DataFrame(np.random.randint(1, 100, size=(5, 3)),
                  columns=['A', 'B', 'C'])

print("Original DataFrame:")
print(df)

# Function to double the values
def double(x):
    return x * 2

# Apply function to DataFrame
result = df.apply(double)

print("\nDataFrame after doubling:")
print(result)
```

```python


# Create DataFrame with 3 columns and 6 rows
df = pd.DataFrame(np.random.randint(1, 100, size=(6, 3)),
                  columns=['A', 'B', 'C'])

print("Original DataFrame:")
print(df)

# Apply lambda function to calculate row-wise sum
df['Total'] = df.apply(lambda x: x['A'] + x['B'] + x['C'], axis=1)

print("\nDataFrame after adding Total column:")
print(df)
```

```python


# Create a Series with 5 random text strings
data = pd.Series(['hello', 'python', 'pandas', 'computer', 'programming'])

print("Original Series:")
print(data)

# Convert all strings to uppercase
result = data.str.upper()

print("\nUppercase Series:")
print(result)
```
